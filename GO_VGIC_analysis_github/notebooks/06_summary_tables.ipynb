# 06 汇总关键结果

打印仓库中所有 25-GO 指标表的核心数值（最佳 MCC 阈值 + AUROC/AUPRC + 5 种子 mean±SD）。


```python
# -*- coding: utf-8 -*-
import csv, os

BASE = os.path.dirname(os.getcwd())

def read_csv(rel):
    with open(os.path.join(BASE, rel), encoding="utf-8") as f:
        return list(csv.DictReader(f))

def best_mcc(rows):
    """返回最佳 MCC 行（thr 列）"""
    best = max(rows, key=lambda r: float(r["MCC"]))
    return best

print("=== 2026 easy 812 (25-GO) 最佳 MCC ===")
for src in ["netgo3", "netgo4", "redundant", "t010", "default"]:
    rows = read_csv(f"results/2026_easy/easy_25go_metrics_{src}.csv")
    b = best_mcc(rows)
    print(f"  {src:10s} best MCC {float(b['MCC']):.4f} @thr={b['thr']}  "
          f"AUROC {float(b['AUROC']):.4f}  AUPRC {float(b['AUPRC']):.4f}")

print("\n=== 5 种子 test 69 (25-GO) mean±SD ===")
rows = read_csv("results/seeds/seed_test_25go_5seeds_metrics_mean_sd.csv")
for r in rows:
    print(f"  {r['source']:7s} thr={r['thr']:4s} MCC {r['MCC_mean±SD']:14s} "
          f"AUROC {r['AUROC_mean±SD']:14s} AUPRC {r['AUPRC_mean±SD']}")

print("\n=== 5 种子 train 274 (25-GO) mean±SD ===")
rows = read_csv("results/seeds/seed_train_25go_5seeds_metrics_mean_sd.csv")
for r in rows:
    print(f"  {r['source']:7s} thr={r['thr']:4s} MCC {r['MCC_mean±SD']:14s} "
          f"AUROC {r['AUROC_mean±SD']:14s} AUPRC {r['AUPRC_mean±SD']}")

print("\n=== 108 共有 entry (25-GO) mean±SD ===")
rows = read_csv("results/common108/train_25go_108_common_metrics_mean_sd.csv")
for r in rows:
    print(f"  {r['source']:7s} thr={r['thr']:4s} MCC {r['MCC_mean±SD']:14s} "
          f"AUROC {r['AUROC_mean±SD']:14s} AUPRC {r['AUPRC_mean±SD']}")

print("\n=== trEMBL Panel A (25-GO) ===")
for r in read_csv("results/trembl/panelA_2023_overlap_25go.csv"):
    print(" ", r)
```
