# 05 电压门控 GO 标签审计（25-GO 集合的确立）

## 目的
从 2026 easy 五个预测源（netgo3 / netgo4 / redundant / t010 / default）逐条解析
**term 含 "voltage"**（不区分大小写）的 GO 标签，确认"无标准标签（GO:0022832 / GO:0005248）但含其他 voltage 标签"的蛋白，
最终确立 25 个 mf 通道活性类 GO 作为统一判定集合（`all_voltage_go_tags.csv`）。

## 结论
- 共发现 42 种 term 含 voltage 的 GO（含标准标签）；排除 cc/bp 类后保留 **25 个 mf 通道活性**标签。
- 关键非标准标签：GO:0005244（voltage-gated monoatomic ion channel activity，netgo4 370/812 蛋白，覆盖最广）。
- 扩展 25-GO 后，NetGO 家族（netgo3/4/redundant）指标显著提升；DeepGO（t010/default）因 GO:0022832 分数恒最高而几乎不变。


```python
# -*- coding: utf-8 -*-
import csv, os, re

BASE = os.path.dirname(os.getcwd())

# 从预测文件解析所有 term 含 voltage 的 GO（以已交付汇总表为例）
# 明细表：results/2026_easy/voltage_nonstandard_go_tags_summary.csv
#          results/2026_easy/voltage_nonstandard_no_std_proteins.csv
summary = {}
with open(os.path.join(BASE, "results", "2026_easy", "voltage_nonstandard_go_tags_summary.csv"), encoding="utf-8") as f:
    for r in csv.DictReader(f):
        summary[(r.get("Source"), r.get("GO_ID"))] = r

print("非标准 voltage GO 汇总条目数:", len(summary))
# 展示前 10 条
for i, ((src, go), r) in enumerate(sorted(summary.items())):
    if i >= 10:
        break
    print(src, go, r.get("GO_Term"), "n=", r.get("n_proteins"))

# 25-GO 集合（最终判定口径）
with open(os.path.join(BASE, "data", "labels", "all_voltage_go_tags.csv"), encoding="utf-8") as f:
    tags = list(csv.DictReader(f))
print("\n25-GO 集合:")
for t in tags:
    print(" ", t["GO_ID"], "-", t["GO_Term"])
```
